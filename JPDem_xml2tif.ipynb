```python
from convert_fgd_dem.converter import Converter
import glob
import os

ziplist = glob.glob("/home/naveen/files/DEMxml/inp/*.zip")

class DummyFeedback:
    def pushInfo(self, msg): print(msg)
    def setProgress(self, value): pass
    def reportError(self, msg): print("ERROR:", msg)
    def isCanceled(self): return False   # Always returns False (never canceled)

feedback = DummyFeedback()

for zipfile in ziplist:
    print(f"Processing ZIP file: {zipfile}")
    # Set the path to your input XML file, ZIP file, or directory containing XML files
    import_path = zipfile

    # Set the output directory and desired output TIFF filename
    output_path = "/home/naveen/files/DEMxml/out"
    file_name = os.path.splitext(os.path.basename(zipfile))[0] + ".tif" 

    # You can specify other options such as EPSG code, RGB conversion, etc.
    converter = Converter(
        import_path=import_path,
        output_path=output_path,
        output_epsg="EPSG:4326",   # Common EPSG for GeoTIFF
        file_name=file_name,
        feedback=feedback,  # Add this line!
    )

    # Run the conversion
    converter.run()
```
